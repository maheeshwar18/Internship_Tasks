```python
# Qno 1: Find the shortest word in a Sentence

# Observe: the given input is a sentence and the expected output is the shortest word in sentence.
# Understand: We have to compare the lengths of all words in the sentence.
# Break: as the input is a sentence, we have to split it to compare length of words.
# Logic: I have to go through len of all words and update when ever i find a shorter word than existing.So i have to use for loop.
# Code: splitting sentence into words ->take first word as shortest -> for every word if current word length < shortest -> shortest= word-> show shortest.
# Test 1

sentence = "Python makes coding easy"
words = sentence.split()
shortest = words[0]

for word in words:
  if len(word) < len(shortest):
    shortest = word
print(shortest)

# Test 2
sentence = "The Filmmaker's Handbook"
words = sentence.split()
shortest = words[0]

for word in words:
  if len(word) < len(shortest):
    shortest = word
print(shortest)
```

```python
# Qno 2: Find the largest number in a List

# Observe: the given input is a list of numbers and the expected output is the largest number.
# Understand: we have to find the largest number without using the max() function.
# Break: take the first number as the largest and compare it with other numbers.
# Logiac: We have to go through all numbers and update whenever we find a greater number. So i have to use a for loop.
# Code: Take the first number as largest -> for every number if current num > existing -> largest = num -> show.
# Test 1

numbers = [12, 45, 7, 89, 23]
largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num
print(largest)

# Test 2
numbers = [-8, -2, -15, -4]
largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num
print(largest)
```

```python
# Qno 15: Create a dictionary of numbers and their even/odd status

# Observe: The input is a list of numbers and the expected output is a dictionary of number's even or odd status.
# Understand: number should be a key and its status should be the value.
# Break: Process each number -> check remainder after division by two -> assign status and assign zero if num is 0.
# Logic: If the remainder is zero, the number is even; otherwise, it is odd.
# Code: Use a for loop and an if num is 0 key should be num and value should be Zero-> elif like wise Even if divisible by 2-> else odd
# Test 1

numbers = [1, 2, 3, 4, 5]
result = {}

for num in numbers:
    if num == 0:
      result[num] = "Zero"
    elif num % 2 == 0:
        result[num] = "Even"
    else:
        result[num] = "Odd"
print(result)

# Test 2
numbers = [-2, -1, 0, 3]
result = {}

for num in numbers:
    if num == 0:
      result[num] = "Zero"
    elif num % 2 == 0:
        result[num] = "Even"
    else:
        result[num] = "Odd"
print(result)
```
