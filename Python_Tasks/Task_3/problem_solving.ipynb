Q1. Find the Frequency of Each Word

Code: Convert the sentence to lowercase and split it into words → create an empty dictionary → for every word, check if it exists in the dictionary → increase its count or add it with count 1 → show the dictionary.

Expected Output:

{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}

{'hello': 2, 'world': 1}

```python

sentence = "python is easy and python is powerful"
words = sentence.lower().split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)

# Additional test case
sentence = "hello world hello"
words = sentence.lower().split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

Q13. Rotate a List to the Right by K Positions

Code:Check if the list is empty → calculate k % len(nums) → take the last K elements using nums[-k:] → take the remaining elements using nums[:-k] → join both parts → show the result.

Expected output:

[4, 5, 1, 2, 3]

[20, 30, 40, 10]

```python

nums = [1, 2, 3, 4, 5]
k = 2

if nums:
    k = k % len(nums)

    if k == 0:
        result = nums.copy()
    else:
        result = nums[-k:] + nums[:-k]
else:
    result = []

print(result)

# Additional test case
nums = [10, 20, 30, 40]
k = 3

if nums:
    k = k % len(nums)

    if k == 0:
        result = nums.copy()
    else:
        result = nums[-k:] + nums[:-k]
else:
    result = []

print(result)
```

Q14. Find the Maximum Sum of Two Consecutive Elements

Code: Add the first two elements → store the sum as max_sum → go through each consecutive pair → calculate current_sum → if current_sum > max_sum, update max_sum → show max_sum.

Expected output:

14

13

```python

nums = [4, 7, 2, 9, 5]
max_sum = nums[0] + nums[1]

for i in range(1, len(nums) - 1):
    current_sum = nums[i] + nums[i + 1]

    if current_sum > max_sum:
        max_sum = current_sum

print(max_sum)

# Additional test case
nums = [1, 2, 10, 3]
max_sum = nums[0] + nums[1]

for i in range(1, len(nums) - 1):
    current_sum = nums[i] + nums[i + 1]

    if current_sum > max_sum:
        max_sum = current_sum

print(max_sum)
```

Q15. Find All Unique Pairs Whose Sum Equals a Target

Code: Create an empty pairs list → compare every pair using nested for loops → check whether their sum equals the target → sort the two numbers → add the pair only if it is not already present → sort the final list → show the pairs.

Expected output:

[[1, 8], [2, 7], [4, 5]]

[[1, 5], [2, 4]]

```python

nums = [2, 4, 3, 5, 7, 8, 1]
target = 9
pairs = []

for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i] + nums[j] == target:
            pair = sorted([nums[i], nums[j]])

            if pair not in pairs:
                pairs.append(pair)

pairs.sort()
print(pairs)

# Additional test case
nums = [1, 2, 3, 4, 5]
target = 6
pairs = []

for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i] + nums[j] == target:
            pair = sorted([nums[i], nums[j]])

            if pair not in pairs:
                pairs.append(pair)

pairs.sort()
print(pairs)
```
