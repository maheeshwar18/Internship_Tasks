```python
# Qno 35
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)

# prediction: [[1,99],[3,4]]
# copy.copy creates a shallow copy where the outer list is new but the nested lists are shared, so the changes will be reflected in a.
```

```python
# Qno 36
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)

# prediction: [[1,2],[3,4]]
# copy.deepcopy creates a copy including the nested lists, so any changes made in copy are not reflected in a.
```

```python
# Qno 37
a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)

# prediction: [[10],[20,30]]
# a.copy creates a shallow copy, so the changes made to the nested list in a will be reflected in b too.
```

```python
# Qno 38
a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)

# prediction: [[10],[20]]
# copy.deepcopy copies everything including lists, so the changes made to a are not reflected in b.
```
